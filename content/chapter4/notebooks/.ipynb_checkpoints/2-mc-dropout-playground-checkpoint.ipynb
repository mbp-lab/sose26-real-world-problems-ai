```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("../data/rural_screening.csv")

print(f"Patients: {len(df)}")
print(f"Disease prevalence: {df['disease'].mean():.1%}")
print(df["disease"].value_counts().rename({0: "No disease", 1: "Disease"}))

fig, ax = plt.subplots(figsize=(8, 6))

for label, color, name in [(0, "steelblue", "No disease"), (1, "crimson", "Disease")]:
    mask = df["disease"] == label
    ax.scatter(
        df.loc[mask, "feature 1"],
        df.loc[mask, "feature 2"],
        c=color,
        label=name,
        alpha=0.5,
        s=15,
        edgecolors="none",
    )

ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
ax.set_title("Rural screening dataset — two classes")
ax.legend(title="True label")
plt.tight_layout()
plt.show()
```

---
#### 🟢 Station 2 — Train a baseline MLP

Build a small vanilla MLP on the two numeric features. Standardize using the training split only, train with early stopping on validation loss, then predict probabilities with `sigmoid(logits)`.

- `STRATIFY = True` — stratified train/val/test splits + class-weighted `BCEWithLogitsLoss`
- `MC_DROPOUT = True` — add dropout layers after each hidden ReLU (active during training as regularization; disabled at prediction time via `model.eval()`)

```python
import random

import numpy as np
import torch
import torch.nn as nn
from sklearn.metrics import accuracy_score, brier_score_loss
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

SEED = 43
STRATIFY = False  # if True: stratified splits + class-weighted BCE loss
MC_DROPOUT = True  # if True: dropout layers for regularization during training
DROPOUT_P = 0.2
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

FEATURE_COLS = ["feature 1", "feature 2"]
TARGET_COL = "disease"
THRESHOLD = 0.5

split_stratify = df[TARGET_COL] if STRATIFY else None

train_val_idx, test_idx = train_test_split(
    df.index,
    test_size=0.2,
    random_state=SEED,
    stratify=split_stratify,
)
train_val_stratify = df.loc[train_val_idx, TARGET_COL] if STRATIFY else None
train_idx, val_idx = train_test_split(
    train_val_idx,
    test_size=0.2,
    random_state=SEED,
    stratify=train_val_stratify,
)

X_train_raw = df.loc[train_idx, FEATURE_COLS].values.astype(np.float32)
X_val_raw = df.loc[val_idx, FEATURE_COLS].values.astype(np.float32)
X_test_raw = df.loc[test_idx, FEATURE_COLS].values.astype(np.float32)
y_train = df.loc[train_idx, TARGET_COL].values.astype(np.float32)
y_val = df.loc[val_idx, TARGET_COL].values.astype(np.float32)
y_test = df.loc[test_idx, TARGET_COL].values.astype(np.float32)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train_raw).astype(np.float32)
X_val = scaler.transform(X_val_raw).astype(np.float32)
X_test = scaler.transform(X_test_raw).astype(np.float32)

input_dim = X_train.shape[1]
print(f"STRATIFY={STRATIFY} | MC_DROPOUT={MC_DROPOUT}" + (f" (p={DROPOUT_P})" if MC_DROPOUT else ""))
print(f"Input dimension (after preprocessing): {input_dim}")
print(f"Train: {len(train_idx)} | Validation: {len(val_idx)} | Test: {len(test_idx)}")
print(f"Train prevalence: {y_train.mean():.1%} | Val: {y_val.mean():.1%} | Test: {y_test.mean():.1%}")


class MLP(nn.Module):
    def __init__(self, input_dim, use_dropout=False, dropout_p=0.2):
        super().__init__()
        layers = [nn.Linear(input_dim, 32), nn.ReLU()]
        if use_dropout:
            layers.append(nn.Dropout(dropout_p))
        layers.extend([nn.Linear(32, 16), nn.ReLU()])
        if use_dropout:
            layers.append(nn.Dropout(dropout_p))
        layers.append(nn.Linear(16, 1))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x).squeeze(-1)


model = MLP(input_dim, use_dropout=MC_DROPOUT, dropout_p=DROPOUT_P)

if STRATIFY:
    n_pos = y_train.sum()
    n_neg = len(y_train) - n_pos
    pos_weight = torch.tensor(n_neg / n_pos)
    criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
    print(f"Class-weighted loss: pos_weight={pos_weight.item():.2f}")
else:
    criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

X_train_t = torch.from_numpy(X_train)
y_train_t = torch.from_numpy(y_train)
X_val_t = torch.from_numpy(X_val)
y_val_t = torch.from_numpy(y_val)


def predict_proba(model, X):
    model.eval()
    with torch.no_grad():
        X_t = torch.from_numpy(X.astype(np.float32)) if isinstance(X, np.ndarray) else X
        logits = model(X_t)
        probs = torch.sigmoid(logits)
    return probs.numpy()


max_epochs = 100
patience = 10
best_val_loss = float("inf")
patience_counter = 0
best_state = None
stopped_epoch = 0

for epoch in range(1, max_epochs + 1):
    model.train()
    optimizer.zero_grad()
    train_loss = criterion(model(X_train_t), y_train_t)
    train_loss.backward()
    optimizer.step()

    model.eval()
    with torch.no_grad():
        val_loss = criterion(model(X_val_t), y_val_t).item()

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = {k: v.clone() for k, v in model.state_dict().items()}
        patience_counter = 0
    else:
        patience_counter += 1
        if patience_counter >= patience:
            stopped_epoch = epoch
            break
else:
    stopped_epoch = max_epochs

if best_state is not None:
    model.load_state_dict(best_state)

print(f"Training stopped at epoch {stopped_epoch} (best validation loss: {best_val_loss:.4f})")

test_probs = predict_proba(model, X_test)
test_preds = (test_probs >= THRESHOLD).astype(int)

print(f"\nTest set (n={len(test_idx)}):")
print(f"  Accuracy (threshold={THRESHOLD}): {accuracy_score(y_test, test_preds):.3f}")
print(f"  Brier score: {brier_score_loss(y_test, test_probs):.4f}")
print(f"  Predicted probabilities: min={test_probs.min():.3f}, max={test_probs.max():.3f}, mean={test_probs.mean():.3f}")
```

---
#### 🟢 Station 3 — Post-hoc recalibration

The validation set was used for early stopping; we now also use it as a **calibration set** to fit temperature scaling and isotonic regression on the MLP's predicted probabilities. Results are evaluated on the held-out test set.

```python
from scipy.optimize import minimize_scalar
from sklearn.calibration import calibration_curve
from sklearn.isotonic import IsotonicRegression

# Validation set = calibration set (not used for training the MLP)
val_probs = predict_proba(model, X_val)

def logits_from_probabilities(probabilities, eps=1e-6):
    probabilities = np.clip(probabilities, eps, 1 - eps)
    return np.log(probabilities / (1 - probabilities))


def probabilities_from_logits(logits):
    return 1 / (1 + np.exp(-logits))


def apply_temperature(probabilities, temperature):
    logits = logits_from_probabilities(probabilities)
    return probabilities_from_logits(logits / temperature)


def fit_temperature(probabilities, y):
    def loss(temperature):
        calibrated = apply_temperature(probabilities, temperature)
        calibrated = np.clip(calibrated, 1e-6, 1 - 1e-6)
        return -np.mean(y * np.log(calibrated) + (1 - y) * np.log(1 - calibrated))

    result = minimize_scalar(loss, bounds=(0.05, 20), method="bounded")
    return result.x


temperature = fit_temperature(val_probs, y_val)
temp_test = apply_temperature(test_probs, temperature)

iso_reg = IsotonicRegression(out_of_bounds="clip")
iso_reg.fit(val_probs, y_val)
iso_test = iso_reg.predict(test_probs)

print(f"Learned temperature: {temperature:.3f}")

methods = [
    ("Original MLP", test_probs),
    ("Temperature scaling", temp_test),
    ("Isotonic regression", iso_test),
]


def referred_disease_pct(y_true, probs, threshold=THRESHOLD):
    referred = probs >= threshold
    if referred.sum() == 0:
        return float("nan")
    return 100 * y_true[referred].mean()


results = pd.DataFrame({
    "Method": [name for name, _ in methods],
    "Brier score": [brier_score_loss(y_test, probs) for _, probs in methods],
    "Accuracy": [
        accuracy_score(y_test, (probs >= THRESHOLD).astype(int))
        for _, probs in methods
    ],
    "Patients referred": [int((probs >= THRESHOLD).sum()) for _, probs in methods],
    "% referred with disease": [
        referred_disease_pct(y_test, probs) for _, probs in methods
    ],
})
display(results.style.format({
    "Brier score": "{:.4f}",
    "Accuracy": "{:.3f}",
    "% referred with disease": "{:.1f}%",
}))

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot([0, 1], [0, 1], "k--", label="Perfect calibration")

colors = ["C0", "C1", "C2"]
for (name, probs), color in zip(methods, colors):
    mean_observed, mean_predicted = calibration_curve(
        y_test, probs, n_bins=10, strategy="quantile"
    )
    ax.plot(mean_predicted, mean_observed, "o-", color=color, label=name)

ax.set_xlabel("Mean predicted risk")
ax.set_ylabel("Observed disease rate")
ax.set_title("Calibration curves (test set)")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.legend()
plt.tight_layout()
plt.show()
```
